## 1. Three Ways to Use LLMs in Coding

Based on the [CodeRefinery lesson](https://coderefinery.github.io/coding-with-ai/), there are three main scenarios for using LLMs in coding, ordered by increasing automation:

![Three scenarios for using LLMs in coding](scenarios.png)

### Scenario I: Chatbots
- **Pros:** Low risk, high control, better understanding of what you're doing
- **Cons:** Required being able to read code, manual copy-pasting (very slow), limited context and context switching between IDE and chatbot

### Scenario II: IDE-integrated assistants
- **Pros:** Seamless workflow, context-aware, huge productivity boost
- **Cons:** Required being able to read code, can be damaging to your code if not properly review, 

### Scenario III: Agentic coding tools
- **Pros:** Do not required being able to read code
- **Cons:** Less control, harder to verify, high risk for unpredictable behavior

### Why do we prefer Scenario II (IDE-integrated assistants)

Scenario II offers the best balance:
- It provides significant productivity gains
- It maintains developer control and understanding
- It fits naturally into existing workflows
- It's appropriate for most research and data analysis tasks

You can see a LLM as an **AI assistant**, not a problem solver. You still need to:
- Define the problem clearly
- Design the algorithm (pseudocode)
- Verify the solution
- Understand what the code does

## 2. The three modes of IDE-integrated assistants

IDE-LLM integration options keep changing over time and differ from one tool to another (Copilot Pro, Claude, Codex, Cursor...). But in general, you will find **three useful modes** in this kind of set-up: **ask**, **agent** and **plan**.

### i. Ask mode

In **ask mode**, the LLM only answers in the chat: it does **not** edit your files. It is the safest mode and the one to use when you want to **understand** something rather than change it:

- explain a piece of code you didn't write (or that the LLM just wrote for you),
- explain an error message,
- ask for alternatives before deciding what to implement.

**Example:** in Class 6, you traced the bug in [`debugging.py`](../06-advanced-python/debugging.py) with `print` statements. Open it and ask:

---

```
Explain what #debugging.py does step by step. The longest heat spell it finds
looks too long compared to the temperature plot. Why? Don't fix the code, just explain.
```

---

Compare the explanation with what you found with print tracing. You can also reference **multiple files** in the same question (generally with `#` or `@` followed by the file name): the LLM reads all of them as context.

> Note: "Don't fix the code, just explain" matters: you want to understand the bug, not just make it disappear.

### ii. Agent mode: the Euler method

In **agent mode**, the LLM can read and **edit your files** directly, and run several steps on its own. It is the mode that saves you the most time, but also the one where you must **review every change** it makes.

In previous classes, I recommended solving this Euler Exercise in 2 steps:
1. Translate plain English tasks into pseudocode
2. Convert pseudocode into real Python code

LLMs can save you lot of time at step 2, with very little implementation risk, given that most of the decisions required to solve the problem have already been made.

As a reminder, here is the exercise from Class 6:

---

We'll examine how forest floors store carbon in dead organic matter (litter).

A classic **one-pool** model ([Olson, 1963](https://doi.org/10.2307/1932179)) balances litterfall input against decomposition:

$$\frac{dC(t)}{dt} = L - k\,C(t)$$

- $C(t)$ = litter carbon stock (Mg C ha $^{-1}$) as a **continuous** function of time $t$, $\frac{dC(t)}{dt}$ is the rate of change
- $L$ = litterfall input (Mg C ha $^{-1}$ yr $^{-1}$)
- $k$ = fractional decomposition rate (yr $^{-1}$)

This is a **first-order ODE**. We can numerically solve it with the **Euler method** (see figure).

At $(t_n, C_n)$, take the local slope $F(t_n, C_n) = L - k\,C_n$ and step forward by $\Delta t$:

$$C_{n+1} = C_n + \Delta t \cdot F(t_n, C_n) = C_n + \Delta t \cdot (L - k\,C_n)$$

![Euler method: continuous C(t) vs discrete C_n steps](../06-advanced-python/img/euler-method.jpg)

Apply Euler to a temperate pine stand with $L = 2.5$, $k = 0.25$ (close to Olson's pine example), starting from a low litter stock after disturbance (2.0). Calculate the Litterfall carbon of the system for the next 20 years and plot your result.

---

Open [`euler_problem.py`](euler_problem.py). It only contains step 1: the tasks in plain English and the pseudocode. We will use the agent mode to translate this pseudocode into Python, then check that the plot matches the Class 6 solution.

> Note: The more relevant context you give (and the less irrelevant context), the better the answer.

### iii. Plan mode

In **plan mode**, the LLM does **not** write code right away: it first proposes a step-by-step **plan** (what to change, in which order, which decisions to make). You review and correct the plan, and only then let the agent implement it.

This is useful for bigger tasks where the decisions have **not** been made yet. In other words, plan mode helps you with step 1 (plain English → pseudocode), but **you** are still the one who must check the decisions.

**Example:** in Class 6, Exercise D asked you to convert **Class 5a** (gridded temperature) or **Class 5b** (survey data) into a `.py` script with `# %%` cells, the same way Class 4 became [`station-processing.py`](../06-advanced-python/station-processing.py). This is a typical multi-step task: many cells to port, figures to save, paths to fix.

Switch to plan mode, reference both files and ask:

---

```
Plan how to convert #05a-geospatial-data.ipynb into a .py script, following the
same structure as #station-processing.py: # %% cells, one cell per step,
figures saved with plt.savefig() / plt.close() instead of plt.show().
Don't write any code yet.
```

---

Before accepting the plan, check:
1. Does it keep the same steps as the notebook (load → select → reduce / mask → exercises)?
2. Are the data paths correct from where the script will be saved?
3. Does every plot end with `plt.savefig(...)` and `plt.close()`, and where are the figures saved?
4. Does it leave out what you don't need (e.g. cells that only display a DataFrame)?

Then let the agent implement it, run the script cell by cell, and compare with the Class 6 solution ([`post-processing.py`](../06-advanced-python/post-processing.py) for 5a, [`survey-processing.py`](../06-advanced-python/survey-processing.py) for 5b).

## 3. AI benchmarks: how relevant are they?

Every new model comes with a press release claiming it beats all others on some benchmark. In practice, these benchmarks have **little relevance** for choosing the tool you will use day to day ([see this paper](https://openreview.net/pdf?id=mdA5lVvNcU)).

Even a carefully designed benchmark like [Humanity's Last Exam](https://lastexam.ai/) (thousands of expert-level questions) has a structural problem: once the questions are public, they can **leak into the training data** of the next models. A model can then score well because it has seen the questions before, not because it can reason about them. (The authors keep a private set of held-out questions precisely to detect this.)

### Cost efficiency

One part of these benchmarks is more interesting: **cost efficiency**. Today, the biggest disparity between models is not their intelligence but the **resources they use** to get there.

![Intelligence Index vs. Cost per Task](intelligence-vs-cost.png)

*Source: [Artificial Analysis](https://artificialanalysis.ai/)*

How to read this graph:
- **y-axis**: the Artificial Analysis Intelligence Index, an average score over a set of benchmark tasks (higher = "smarter")
- **x-axis**: the average cost (USD) to run one task, on a **log scale** (each tick is a multiplication, not an addition)
- **Green quadrant**: the most attractive models, high score at a low cost
- **Pareto line** (dotted): the best score you can get for a given cost. Models below/right of it are beaten by a cheaper or smarter model

The scores of the top models sit within a few points of each other (≈ 50–60), while their cost per task varies by about **10×** (≈ \$1 to \$13).
