## 1. Three ways to use LLMs in Coding

The key, when using LLMs for coding, lies in striking the right balance between risks and productivity: the more we automate, the faster we become, but the less control and understanding we retain.

[CodeRefinery](https://coderefinery.github.io/coding-with-ai/introduction/) describes three main scenarios for using LLMs in coding, ordered by increasing automation:

![Three scenarios for using LLMs in coding](scenarios.png)

### Scenario I: Chatbots
- Example: this is how most of us started using LLMs: asking a question to ChatGPT in a browser tab, then copy-pasting the code into our editor
- Pros: Low risk, high control, better understanding of what you're doing
- Cons: Requires being able to read code, manual copy-pasting (very slow), limited context between IDE and chatbot

### Scenario II: IDE-integrated assistants
- Examples: Copilot Pro or Claude extension in VS Code: the LLM lives inside your editor, sees your files and proposes changes that you review there
- Pros: Seamless workflow, context-aware, huge productivity boost
- Cons: Requires being able to read code, can be damaging to your code if not properly reviewed, small risk for unpredictable behavior

### Scenario III: Agentic coding tools
- Examples: Claude Code or Codex app: you describe the task, the agent edits and runs the code on its own, and you don't necessarily see the code it writes
- Pros: Does not require being able to read code
- Cons: Less control, harder to verify, high risk for unpredictable behavior

### Why do we prefer Scenario II (IDE-integrated assistants)

Scenario II offers the best balance:
- It provides significant productivity gains...
- ... while maintaining developer control and understanding
- It fits naturally, almost organically, into existing workflows

You can see an LLM as an AI assistant, not a problem solver. You still need to:
- Define the problem clearly
- Design the algorithm
- Verify the solution
- Understand what the code does

## 2. How to use IDE-integrated assistants

IDE-LLM integration options keep changing over time and differ from one tool to another (Copilot Pro, Claude, Codex, Cursor...). But in general, you will find three useful modes in this kind of set-up: ask, agent and plan.

### i. Ask mode

In ask mode, the LLM only answers in the chat: it does not edit your files. It is the safest mode and the one to use when you want to understand something rather than change it:

- Explain a piece of code you didn't write (or that the LLM just wrote for you)
- Explain an error message
- Ask for alternatives before deciding what to implement

Example: in Class 6, you traced the bug in [`debugging.py`](../06-advanced-python/debugging.py) with `print` statements. Instead of doing that, open it and use the Copilot Pro ask mode to ask:

You can reference one or more files by typing `@` followed by the file name, and the LLM will read it as context. To point it more precisely at where it should work, you can also select specific lines with your mouse, which is especially useful for large files.

### ii. Agent mode

In agent mode, the LLM can read and edit your files directly, and run several steps on its own. It saves you a lot of time, but it is highly recommended to review every change it makes.

In previous classes, I recommended solving this Euler Exercise in 2 steps:
1. Translate plain English tasks into pseudocode
2. Convert pseudocode into real Python code

You can let the LLM do the second step, with very little implementation risk, given that most of the decisions required to solve the problem have already been made.

Open [`euler_problem.py`](euler_problem.py) which contains the tasks in plain English and the pseudocode. Then use the agent mode to translate this pseudocode into Python, then check that the plot matches the Class 6 solution.

#### Reviewing the agent's changes

Now that the changes have been made directly to your code, it’s your responsibility to review them, making sure they make sense to you and produce good results. Sometimes, agents may commit the changes they have made before you have even had a chance to review them. If this happens, adjust your prompt accordingly to prevent this!

Accept the changes one by one. The agent can change one or several files. Its changes are not final yet: they appear in your editor as a diff (added lines in green, removed lines in red). Use the Keep / Undo bar to accept or reject each change.

![Keep / Undo bar to accept or reject each change](../img/keep_undo.png)

Every single line must be clear to you. Ask yourself if you know every functions used. This is the perfect moment to switch to ask mode: select the line and ask, e.g. *"What does `np.zeros(NSteps + 1)` do here, and why `+ 1`?"*. Only press Keep once the answer makes sense and produce the desired outputs.


### iii. Plan mode

In plan mode, the LLM does not write code right away: it first proposes a step-by-step plan (what to change, in which order, which decisions to make). You review and correct the plan, and only then let the agent implement it. This is useful for bigger tasks where the decisions have not been made yet.

Example: in Class 6, Exercise D asked you to convert Class 5a (gridded temperature) or Class 5b (survey data) into a `.py` script with `# %%` cells, the same way Class 4 became [`station-processing.py`](../06-advanced-python/station-processing.py). This is a typical multi-step task: many cells to port, figures to save, paths to fix.

Switch to plan mode, reference both files and ask:

> Agents don't only write code: they can also run commands in a terminal (e.g. `python 05a-geospatial-data.py`), read the output or the error message, fix the code and run it again, until it works. This is very useful to catch errors (wrong paths, missing imports, typos) before you even look at the result.
>
> Most tools run these commands in a sandbox: an isolated environment where the agent can try things out without touching the rest of your computer. Generally, outside the sandbox, the agent will ask for your permission before running a command.
>
> Keep in mind that *"it runs without errors"* does not mean *"it is correct"*: you still need to review the code and check the outputs created.

## 3. Context and prompting

Each LLM you use (paid or free) comes with a maximum usage, reset daily, weekly or monthly depending on the provider. Obviously, the more requests you send, the more usage you consume. But not every request costs the same: what you pay for is mostly the context, i.e. everything the LLM has to read before answering:
- your prompt,
- the files you reference with `@` and the lines you selected,
- sometimes the file currently open in your editor and all files in your repo,
- the whole chat history of the current conversation,
- context files (`AGENTS.md`, `CLAUDE.md`...)

The more context, the more usage, and (beyond a certain point) the worse the answer: the important information gets diluted in the rest (usually referred to as ["context rot"](https://www.trychroma.com/research/context-rot)).

Prompting tips change from one model to the next, and even between versions of the same model. So the best general advice is to search for "*Prompting best practices for [insert your favourite model]*" and read the official guide of the provider. 

However, you will notice that most of these guides target either chatbots (Scenario I) or agentic / "vibe" coding (Scenario III). Very few guides are written for what we do here (Scenario II): you make the scientific and algorithmic decisions, and the LLM speeds up the typing.

Still, I can try to give you some tips that work for most set-ups today, but this may turn into the complete opposite in a few months!

1. Specify your context: reference only the files you need with `@` (if you want to exclude files to be read by an agent, check [here](https://github.com/yjcho9317/aiignore-cli/blob/main/docs/test-report.md))
2. Show an example: providing an example is one of the most reliable ways to steer format, style and structure. E.g. I like to give a kind of plot from another paper I want to reproduce with my data and ask what should be needed (but [some Anthropic engineers think the opposite](https://simonwillison.net/2026/Jul/21/cat-and-thariq/))
3. Ask for questions before code: end your prompt with "*If something is ambiguous, ask me before writing code*". A question from the LLM is a decision you should make, instead of a decision silently made for you
4. Start a new chat for each new task

Also, note that a [recent study on coding agents](https://arxiv.org/abs/2602.11988) found that instruction files (`CLAUDE.md`, `AGENTS.md`...) do not generally improve task success.

## 5. AI benchmarks: how relevant are they?

Every new model comes with a press release claiming it beats all others on some benchmark. In practice, these benchmarks have little relevance for choosing the tool you will use day to day ([see this paper](https://openreview.net/pdf?id=mdA5lVvNcU)).

Only carefully designed benchmarks like [Humanity’s Last Exam](https://lastexam.ai/) (thousands of expert-level questions not release to the public) are harder to game. But they do not test coding.

### Cost efficiency

One part of these benchmarks is more interesting: cost efficiency. Today, the biggest disparity between models is not their intelligence but the resources they use to get there.

You can see on the graph below that while the y-axis (intelligence benchmark) is linear and the scores of the top models sit within a few points of each other (≈ 50–60), the x-axis (cost) is on a log scale and their cost per task varies by about 10× (≈ \$1 to \$13)!
