## 2. Comments

A **comment** is a text that Python **does not execute**. It is there for you (and for others) to explain what the code is doing.

Comments are extremely important. In a few weeks, you will not remember why you wrote a specific line. A short comment can save you (and your colleagues) a lot of time.

In Python, a comment starts with the symbol `#`. Everything after `#` on that line is ignored by Python.

You can write a comment:

- **on its own line**, to explain a block of code
- **at the end of a line**, to explain that specific instruction


````python
# This is a comment
print("Hello Future Forests!") # This is another comment
````

Try to run the cell above: only `Hello Future Forests!` is printed. The comments are ignored.

## Exercise B

Marta has an automatic water fountain for her cat. Before going for a long run, she needs to fill the reservoir with exactly enough water to last until she comes back.

The cat's water consumption depends on how long Marta is away:

- for every complete hour, the cat needs 100 mL of water
- for every additional minute, the cat needs 2 mL of water (see after why)

Marta's next run will last 932 minutes, write a Python program that calculates how much water she should put in the reservoir.

````python
# Solution
BreakMins = 932
MinInOneHour = 60
NbHours = BreakMins // MinInOneHour
print("Number of hours:", NbHours)
NbExtraMin = BreakMins % MinInOneHour
print("Number of extra minutes:", NbExtraMin)
TotalWater = NbHours * 100 + NbExtraMin * 2
print("Total water needed:", TotalWater, "ml")
````

### i. Why use Numpy arrays?

Python lists are not ideal for working with large amounts of **numerical data** (temperatures, measurements, statistics...). Numpy arrays are better because they are **faster** when working with many numbers and come with useful functions.

To create a Numpy array from a Python list, import the `numpy` library and use the function `np.array`:


````python
import numpy as np
````

````python
TempStationsArr = np.array(TempStations)
````

`TempStations` is now a Numpy array (`numpy.ndarray`). From this point on, we will work with arrays rather than Python lists.

## Exercise D (Bonus)

Before planting a new site, you want to know how the rainfall is distributed over the year. The meteorological station next to the site gives you the monthly precipitation (in mm) of last year, from January to December:

76.2, 63.5, 70.1, 51.3, 66.4, 71.8, 74.0, 79.3, 68.6, 75.2, 77.9, 81.0

Write code that

- stores the month names (`"Jan"`, `"Feb"`, ...) in a `list` and the precipitation values in a Numpy array.
- selects the spring months (March, April, May), the summer months (June, July, August) and the autumn months (September, October, November) with slicing, and stores them in three new arrays. Computes and display the average monthly precipitation of each season (the `numpy` library has a function for that, Google "numpy mean" and watch the syntax!)
- computes the total annual precipitation (Google "numpy sum") and the share (in %) of the rain that falls during the growing season, from April to September. Display this share rounded to the nearest integer.

Tip: for slicing, remember that the start is included but the stop is excluded.

````python
# Solution
MonthNames = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
PrecipMonths = [76.2, 63.5, 70.1, 51.3, 66.4, 71.8, 74.0, 79.3, 68.6, 75.2, 77.9, 81.0]
PrecipMonthsArr = np.array(PrecipMonths)

# Slicing: start is included, stop is excluded
PrecipSpring = PrecipMonthsArr[2:5]   # Mar, Apr, May
PrecipSummer = PrecipMonthsArr[5:8]   # Jun, Jul, Aug
PrecipAutumn = PrecipMonthsArr[8:11]  # Sep, Oct, Nov
AvgPrecipSpring = np.mean(PrecipSpring)
AvgPrecipSummer = np.mean(PrecipSummer)
AvgPrecipAutumn = np.mean(PrecipAutumn)
print("Spring mean:", AvgPrecipSpring)
print("Summer mean:", AvgPrecipSummer)
print("Autumn mean:", AvgPrecipAutumn)

# Growing season: Apr (index 3) to Sep (index 8), so the stop is 9
PrecipGrowSeason = np.sum(PrecipMonthsArr[3:9])
TotalPrecipYear = np.sum(PrecipMonthsArr)
ShareGrowSeason = PrecipGrowSeason / TotalPrecipYear * 100
print("Share of precipitation during the growing season: (in %)", round(ShareGrowSeason))
````
