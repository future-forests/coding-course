## 3. Comments

A **comment** is a text that Python **does not execute**. It is there for you (and for others) to explain what the code is doing.

Comments are extremely important. In a few weeks, you will not remember why you wrote a specific line. A short comment can save you (and your colleagues) a lot of time.

In Python, a comment starts with the symbol `#`. Everything after `#` on that line is ignored by Python.

You can write a comment:

- **on its own line**, to explain a block of code
- **at the end of a line**, to explain that specific instruction


````python
# This is a comment on its own line
print("Hello Future Forests!")   # This comment explains the line next to it
````

Try to run the cell above: only `Hello Future Forests!` is printed. The comments are ignored.

> **Tip:** Use comments to explain *why* something is done, not to repeat what the code already says. `# Print a message` is not useful.

## Exercise B

Marta has an automatic water fountain for her cat. Before going for a long run, she needs to fill the reservoir with exactly enough water to last until she comes back.

The cat's water consumption depends on how long Marta is away:

- for every complete hour, the cat needs 100 mL of water
- for every additional minute, the cat needs 2 mL of water (see after why)

Marta always knows the exact duration of her run in minutes.

Her next run will last 932 minutes, but she wants an algorithm that works for any duration.

Write a Python program that calculates how much water she should put in the reservoir.

Tip: Your program should determine how many complete hours are contained in the long run and how many minutes remain after those complete hours. It should then use these values to calculate the total amount of water required.

(More context about the complex numbers: Nobody knows why. Marta suspects that the cat possesses some kind of mystical sixth sense and somehow knows, with supernatural precision, when she is about to come home. As soon as the run enters its final few minutes, the cat becomes increasingly thirsty, tense, and dramatically dissatisfied with the amount of water in his fountain.)


````python
BreakMins = 932

# Solution
MinInOneHour = 60
NbHours = BreakMins // MinInOneHour
print("Number of hours:", NbHours)
NbExtraMin = BreakMins % MinInOneHour
print("Number of extra minutes:", NbExtraMin)


TotalWater = NbHours * 100 + NbExtraMin * 2
print("Total water needed:", TotalWater, "ml")
````

### i. Why use Numpy arrays?

Python lists are not ideal for working with large amounts of **numerical data** (temperatures, measurements, statistics...). Numpy arrays are better for science because they are **faster** when working with many numbers and come with useful functions.

To create a Numpy array from a Python list, import the `numpy` library and use the function `np.array`:


````python
import numpy as np
````

````python
TempStations = np.array(TempStations)
````

`TempStations` is now a Numpy array (`numpy.ndarray`). From this point on, we will work with arrays rather than Python lists.

````python
type(TempStations)
````

````python
BlankArray = np.zeros(10)       # create empty arrays
print(BlankArray)
````

````python
print(np.size(TempStations))    # returns size
````

### ii. Tools to select elements in arrays

Coders and scientists typically work with large datasets that are translate into arrays. One challenge is to select and extract appropriate data from them.

There are **two different tools** you can use to select specific values from an array.

#### a. Indexing: Selecting single elements


````python
FirstTemp = TempStations[0]
print(FirstTemp)
````

⚠️ **Note:** Python uses **zero-based numbering**, so the **first element is indexed with 0** and not 1. ⚠️

To select from the end of the array, you can use negative indices:


````python
TempStations[-1]  # Last element
````

Try to select the second element of `TempStations`


#### b. Slicing: Selecting several elements

We can also select several elements with the slice notation, marked by the **colon character :**

```py
Array[start:stop]
```

**Note:** the start is included, but the stop is excluded (similar to the `range()` function).

To select all values from the second to the fifth element of `TempStations`, type


````python
TempStations[1:5]
````
