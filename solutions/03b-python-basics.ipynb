## 1. Repeating tasks with `for` loop

Imagine you have to print every number in a `list`, you *could* do:

````python
odds = [1, 3, 5]

print(odds[0])
print(odds[1])
print(odds[2])
````

This is a terrible approach for two reasons:
1. Not scalable: Printing hundreds of elements manually is impractical
2. Hard to maintain: Changing what you want to, would require editing multiple lines of code

That's why we use a `for` loop to repeat or iterate the same task a number of times:

````python
odds = [1, 3, 5]

for num in odds:
    print(num)
````

The general form of a loop is:

```py
for iterator in list:
    # do things using iterator, such as print
```

where the `iterator` is a variable that goes through a `list` (**don't forget the `:` symbol at the end**).

Note that the value of the iterator still exists after the loop finishes:

````python
print(num)
````

Try to print all the element of this `NumPy array`:

````python
TempStations = np.array([10.5, 9.0, 12.2, 14.9, 15.5, 8.7])

````

It also work with a `list` of `strings`:

````python
names = ['Curie', 'Darwin', 'Turing']

for name in names:
    print(name)
````

Loops can also be used to **update another variable** step by step (here we count how many names we have):

````python
length = 0 # "initialize a variable"

for name in names: # note: here name is not used
    # print("next iteration starts, length ==", length)

    length = length + 1

    # print("end of iteration, length ==", length)

print('There are', length, 'names in the list.')
````

### Using `enumerate` in a `for` loop

Often you need both the index and the value (for example, "station 2 is at -9.2 °C"). **`enumerate`** yields a pair `(index, value)` as you loop:

````python
List = [10, 20, 30]

for index, value in enumerate(List):
    print(index, value)
````

Try to print each temperature in `TempStations` together with its station number (starting at 0), using `enumerate`:

````python
TempStations = np.array([10.5, 9.0, 12.2, 14.9, 15.5, 8.7])

# Solution
for i, TempValue in enumerate(TempStations):
    print(i, TempValue)
````

**Note**: Python relies on indentation (space at the beginning of a line) to limit the application of a loop:

````python
# Uncomment lines below
for i, value in enumerate([0, 2, 4, 6, 8]):
    print(i, value)
````

## Exercise A

This exercise is a copy of Class 3a Exercise A, except that you must use a `NumPy array` and a `for` loop this time.

To find the best location to plant a temperature-sensitive species in your garden, you measure the temperature at five different locations. The numbers you read on your thermometer are: 20.4°C, 20.8°C, 21.1°C, 20.3°C, and 21.2°C. Calculate the average temperature of your garden.

Your American cousin, an avid gardener, comes to visit you and doesn't understand degrees Celsius. Convert each of your temperature measures to degrees Fahrenheit and calculate the average.


````python
import statistics as stat

# Solution 

# Tasks in plain English
# 1. stores these numbers (20.4°C, 20.8°C, 21.1°C, 20.3°C, and 21.2°C)
# 2. compute the temperature average of these number and display it
# 3. convert all temperature measures to degrees Fahrenheit
# 4. compute the average of the converted numbers and display it

# Pseudocode
# 1. Create an array with the numbers 20.4°C, 20.8°C, 21.1°C, 20.3°C, and 21.2°C
# 2. Use stat.mean and print()
# 3. Create a new empty array of same size (calculate size with np.size) 
#    Loop with enumerate through the Celsius array [i, TempC]
#    Fahr array[i] = TempC * 9/5 + 32
# 4. Same as 2. but with Fahr array

TempGardenC = np.array([20.4, 20.8, 21.1, 20.3, 21.2])

AvgTempC = stat.mean(TempGardenC)
print('Average (°C):', AvgTempC)

NObs = np.size(TempGardenC)
TempGardenF = np.zeros(NObs)  

# fill element by element in the loop below
for i, TempC in enumerate(TempGardenC):
    TempGardenF[i] = TempC * 9 / 5 + 32  # °C → °F (same as Class 3a)
    
AvgTempF = stat.mean(TempGardenF)
print('Average (°F):', AvgTempF)
````

### Tip to solve complex problem in programming

When a problem looks too big to code directly, don't start by typing Python:

1. **Translate the problem into plain English tasks.** Read the question and break it down into a numbered list of small, simple steps (*"store the numbers"*, *"compute the average"*, *"convert each value"*...). At this stage, don't think about Python at all: you just need to understand **what** has to be done, and in which order.
2. **Convert each task into pseudocode.** Pseudocode is a description of your program written in a mix of plain language and programming logic (variables, loops, conditions), but without worrying about the exact syntax of a specific language.
3. **Convert the pseudocode into real code.** Each line of pseudocode now becomes one (or a few) lines of Python. Since the logic is already worked out, you only have to focus on the syntax. If you get stuck on one step, you know exactly what to search for.

`Booleans` represent yes/no questions: is something true, or not?

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |

> **Reminder:** `=` assigns a value to a variable (as you saw with `TempStat1 = 10.5`). The real "equals to" symbol is `==` (two equal signs).

When you use a comparison operator, Python returns `True` or `False`.

````python
print(TempStat2 > TempStat1)
print(TempStat2 < TempStat1)
````

### i. `if` condition

These `booleans` can be used with `if` statements to **make decisions in your code**. It will execute the code in the statement only when the expression is true:

```py
if boolean:
    statement
```


````python
if TempStat2 > TempStat1:
    print(TempStat2, 'is higher than', TempStat1)
````

`TempStat2 > TempStat1` is the boolean, the result can be only True or False. You can also use a variable like this:

````python
TwoWarmerThanOne = TempStat2 > TempStat1

if TwoWarmerThanOne:
    print(TempStat2, 'is higher than', TempStat1)
````

You can add `else` to specify what should happen when the `if` condition is `False`.

```py
if expression:
    statement_if_true
else:
    statement_if_false
```


````python
if TempStat1 > TempStat2:
    print(TempStat2, 'is higher than', TempStat1)       # this is not executed
else:
    print(TempStat2, 'is not higher than', TempStat1)   # this is executed
````

You can also combine a `for` loop with an `if` condition


````python
TempStations = np.array([10.5, 9.0, 12.2, 14.9, 15.5, 8.7])

for TempValue in TempStations:
    if TempValue > 10:
        print(TempValue, '°C is above 10°C')
````

## Exercise F (Bonus)

You are trying to get new very temperature-humidity-dependent plants in your appartment. According to the literature, conditions become risky when the air temperature rises above 28°C **or** relative humidity falls below 40%.

You measure both variables in your five rooms, this time at four times of day (morning, afternoon, evening, and night). The eight arrays are given in the next cell.

Write a code that give the risk in each time of day (morning, afternoon, evening, and night) in %, and calculate the best room to take.

````python
TempMorning = np.array([18.2, 19.1, 18.7, 19.4, 18.0]) # [Room 1, 2, 3, 4, 5]
TempAfternoon = np.array([28.3, 29.1, 27.4, 26.8, 30.2])
TempEvening = np.array([24.1, 25.8, 26.6, 23.9, 26.1])
TempNight = np.array([16.4, 17.2, 16.8, 17.5, 16.1])
HumMorning = np.array([78.0, 75.0, 45.0, 37.0, 45.0])
HumAfternoon = np.array([42.0, 35.0, 38.0, 48.0, 33.0])
HumEvening = np.array([55.0, 48.0, 39.0, 58.0, 45.0])
HumNight = np.array([30.0, 31.0, 43.0, 51.0, 28.0])

# Solution
CountMorning = 0
CountAfternoon = 0
CountEvening = 0
CountNight = 0

NRooms = np.size(TempMorning)
RiskPerRoom = np.zeros(NRooms)  # how many time slots are risky per room

for i, Temp in enumerate(TempMorning):
    if TempMorning[i] > 28 or HumMorning[i] < 40:  # hot OR dry
        CountMorning += 1
        RiskPerRoom[i] += 1
    if TempAfternoon[i] > 28 or HumAfternoon[i] < 40:
        CountAfternoon += 1
        RiskPerRoom[i] += 1
    if TempEvening[i] > 28 or HumEvening[i] < 40:
        CountEvening += 1
        RiskPerRoom[i] += 1
    if TempNight[i] > 28 or HumNight[i] < 40:
        CountNight += 1
        RiskPerRoom[i] += 1

print('Risk morning:', CountMorning / NRooms * 100, '%')
print('Risk afternoon:', CountAfternoon / NRooms * 100, '%')
print('Risk evening:', CountEvening / NRooms * 100, '%')
print('Risk night:', CountNight / NRooms * 100, '%')
print('Best room:', np.argmin(RiskPerRoom))  # fewest risky periods
````

## 3. Vectorisation vs loops

Loops are very flexible, but they are often slower and require more code, making them less readable and harder to maintain.

With `NumPy`, we can apply an operation to an entire array at once through **vectorisation**. This results in code that is shorter, easier to read, and faster to execute. As a general rule, it is recommended to minimize explicit loops and make use of vectorised operations whenever possible.

### i. Math operations

Vectorization greatly simplifies array manipulation with math operations. Imagine that every temperature requires a +0.5°C shift (for example, a calibration adjustment):

````python
TempC = np.array([19.5, 20.0, 21.3, 22.1, 20.4])

# loop
TempSize = np.size(TempC)
TempCorrectedC = np.zeros(TempSize)

for i, Temp in enumerate(TempC):
    TempCorrectedC[i] = Temp + 0.5
    
print(TempCorrectedC)

# vectorisation
TempCorrectedC = TempC + 0.5
print(TempCorrectedC)
````

You can also work with multiple arrays to define a new one. E.g. with a substraction:

````python
TempMod = np.array([19.5, 20.0, 21.3, 22.1, 20.4])
TempObs = np.array([18.2, 20.3, 20.0, 24.4, 19.8])

# loop
TempSize = np.size(TempMod)
TempDiffC = np.zeros(TempSize)

for i, Temp in enumerate(TempMod):
    TempDiffC[i] = TempMod[i] - TempObs[i]
    
print(TempDiffC)

# vectorisation
TempDiffC = TempMod - TempObs
print(TempDiffC)
````

### ii. Logical operations (filtering)

You can create a `Boolean` using a `list`, that's what we can call a `mask`. For example, `TempC > 21.0` does not return a single yes/no answer, but returns one `True` or `False` per element:

````python
TempC = np.array([19.5, 20.0, 21.3, 22.1, 20.4, 24.8])

HotMask = TempC > 21.0
print(HotMask)
````

Writing `TempC[HotMask]` keeps only the temperatures where the condition holds (here, strictly above 21°C):

````python
HotTempC = TempC[HotMask]
print(HotTempC)
````

As you can see, this "rough" filtering returns an array of a different size from the original one. If you want to keep the same size, the function `np.where` can be used:

````python
HotTempsC = np.where(TempC > 21.0, TempC, np.nan)
print(HotTempsC)
````

### iii. Combining masks: `&` and `|`

`and`/`or` only work between two single Booleans. To combine two `masks` element by element, use `&` (AND) and `|` (OR) instead and use parenthesis like this:

````python
TempC = np.array([19.5, 20.0, 21.3, 22.1, 20.4])
HumPct = np.array([65.0, 48.0, 52.0, 38.0, 70.0])

 # True where BOTH are True
HotAndDryMask = (TempC > 21.0) & (HumPct < 50.0) 
print('Hot AND dry:', HotAndDryMask)
print('Hot and dry temperatures:', TempC[HotAndDryMask])

# True where AT LEAST ONE is True
HotOrDryMask = (TempC > 21.0) | (HumPct < 50.0)   
print('Hot OR dry:', HotOrDryMask)
print('Hot and dry temperatures:', TempC[HotOrDryMask])
````

## Exercise H (Bonus)

During a forest inventory in the Black Forest, you measured the diameter at breast height (`DBH`, in cm) and the height (in m) of the 15 trees of the plot. The two arrays are given in the next cell.

Solve every task with vectorisation:

1. Calculate the stem volume (m³) of each tree with the simplified formula below, where the form factor `f = 0.5` accounts for the tapering of the stem:

   $$V = f \cdot \pi \cdot \left(\frac{DBH}{200}\right)^2 \cdot Height$$

   (use `np.pi`)
2. Calculate the total volume of the plot (all trees combined).
3. A tree can be harvested when it is **mature**, i.e. when its `DBH` is at least 35 cm **and** its height is at least 25 m. Give the volume of each mature tree and the total number of mature trees.

````python
DBH = np.array([42.1, 18.5, 36.8, 27.3, 51.0, 33.4, 39.7, 22.0, 45.3, 30.1, 37.9, 15.8, 48.6, 35.2, 29.4])  # cm
Height = np.array([31.2, 15.4, 27.8, 22.1, 34.5, 26.0, 24.3, 18.7, 32.0, 23.9, 28.4, 13.2, 33.1, 24.8, 21.5])  # m

# Solution
NTrees = np.size(DBH)

# 1. Stem volume of every tree at once (vectorised math, no loop)
FormFactor = 0.5
Volume = FormFactor * np.pi * (DBH / 200) ** 2 * Height 
print('Volume per tree (m3):', np.round(Volume, 2))

# 2. Total volume of the plot
TotalVolume = np.sum(Volume)
print('Total volume (m3):', TotalVolume)

# 3. Mature trees: both conditions must hold -> '&' between parentheses
MatureMask = (DBH >= 35) & (Height >= 25)
print('Volume of mature trees (m3):', Volume[MatureMask])
NMature = np.sum(MatureMask)  # np.sum counts the True values (True = 1, False = 0)
print("Total number of mature trees:", NMature)
````
