## Exercise A

This exercise is a copy of Class 3a Exercise A, except that you must use a NumPy array and a `for` loop this time.

To find the best location to plant a temperature-sensitive species in your garden, you measure the temperature at five different locations. The numbers you read on your thermometer are: 20.4°C, 20.8°C, 21.1°C, 20.3°C, and 21.2°C. Write code that (i) stores these numbers in an array and (ii) stores a variable that gives you the temperature average of your garden. Display the average.

Your American cousin, an avid gardener, comes to visit you and doesn't understand degrees Celsius. Convert all temperature measures to degrees Fahrenheit (you have to define a new array and use a for loop). Display the average in degrees Fahrenheit.


````python
import statistics as stat

# Solution
TempGardenC = np.array([20.4, 20.8, 21.1, 20.3, 21.2])

AvgTempC = stat.mean(TempGardenC)
print('Average (°C):', AvgTempC)

NObs = np.size(TempGardenC)
TempGardenF = np.zeros(NObs)  

# fill element by element in the loop below
for i in range(0, NObs):
    TempGardenF[i] = TempGardenC[i] * 9 / 5 + 32  # °C → °F (same as Class 3a)
    print(TempGardenF[i], '(°F)')
    
AvgTempF = stat.mean(TempGardenF)
print('Average (°F):', AvgTempF)
````

### Very useful tip

A common way to use a `for` loop is to go through an array **directly**, without `range()`.

In Exercise A, we looped over the **indices** (0, 1, 2, ...) because we needed the position of each element to fill another array (`TempGardenF[i]`).

When you only need the **values**, you can loop over the array itself:


````python
for TempValue in TempStations:
    print(TempValue)
````

Here, `TempValue` is not the position in the array. It is each value, one after the other. Python gives you every element, from the first to the last. This is shorter, and it is the usual way to write a loop when you want to print the values, add them up, or convert them one by one.

**Tip:** It's also much better to not use `i` for the iterator of this loop. `i`, `j`... should be reserved for indices.

Try to redo the same exercise than before but just by printing the Fahrenheit values, without using `range()` and calculating the averages (it's possible in two lines):

````python
# Solution
# Values only: no index needed when we just convert and print each reading
for TempValue in TempGardenC:
    print(TempValue * 9 / 5 + 32, '(°F)')
````

**To sum-up**: Use `for i in range(...)` when you need the **index**. Use `for value in array` when you only need the **values**.

## Exercise B

After the 2018–2020 droughts killed large areas of spruce, a survey asked private forest owners in the Black Forest whether they would replace spruce with mixed stands. They responded to the statement "I plan to replant with mixed species" on a Likert scale. Each answer is stored as text, one of:

`"strongly disagree"`, `"disagree"`, `"neutral"`, `"agree"`, `"strongly agree"`

Use a dictionary to give each label a score from 1 (strongly disagree) to 5 (strongly agree). Then convert every answer in the list below to its score and calculate the mean.

````python
Answers = ['agree', 'neutral', 'agree', 'disagree', 'strongly agree', 'agree', 'neutral', 'agree']

# Solution
LikertMap = {'strongly disagree': 1, 'disagree': 2, 'neutral': 3, 'agree': 4, 'strongly agree': 5}
AnszersSize = len(Answers)
Scores = np.zeros(AnszersSize) 

 # Convert to numberic score using loop index and Likert scale dict
for i in range(AnszersSize):
    Scores[i] = LikertMap[Answers[i]]

MeanSupport = np.mean(Scores)
print('Scores:', Scores)
print('Mean support (1-5):', MeanSupport)
````

## 2. Making decisions with `if` condition


````python
TempStat1 = 10.5
TempStat2 = 13.0
````


**New variable:** `Boolean` stores only two possible values - True or False:


````python
StationActive = True
print(StationActive)
````

`Booleans` represent yes/no questions: is something true, or not? They can also be the result of a `comparison` between two values. To compare values, Python uses `comparison operators`:

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |

> **Reminder:** `=` assigns a value to a variable (as you saw with `TempStat1 = 10.5`). The real "equals to" symbol is `==` (two equal signs).

When you use a comparison operator, Python returns `True` or `False`.

Here is an example:


````python
print(TempStat2 > TempStat1)
````

You can define a Boolean variable in a similar way

````python
TwoWarmerThanOne = TempStat2 > TempStat1
print(TwoWarmerThanOne)
````

### i. `if` condition

These `booleans` can be used with `if` statements to **make decisions in your code**. It will execute the code in the statement only when the expression is true:

```py
if boolean:
    statement
```


````python
if TempStat2 > TempStat1:
    print(TempStat2, 'is higher than', TempStat1)
````

`TempStat2 > TempStat1` is the boolean, the result can be only True or False. You can also use a variable like this:

````python
TwoWarmerThanOne = TempStat2 > TempStat1
if TwoWarmerThanOne:
    print(TempStat2, 'is higher than', TempStat1)
````

You can add `else` to specify what should happen when the `if` condition is `False`.

```py
if expression:
    statement_if_true
else:
    statement_if_false
```


````python
if TempStat1 > TempStat2:
    print(TempStat2, 'is higher than', TempStat1)
else:
    print(TempStat2, 'is not higher than', TempStat1)
````

You can also combine a `for` loop with an `if` condition


````python
for TempValue in TempStations:
    if TempValue > 10:
        print(TempValue, '°C is above 10°C')
````

## Exercise E (Bonus): The Fibonacci sequence

The **Fibonacci sequence** is a famous series of numbers. It starts with `0` and `1`, and each next number is the **sum of the two previous ones**:

`0, 1, 1, 2, 3, 5, 8, 13, 21, ...`

Write code that prints the **first 20 Fibonacci numbers**


````python
# Solution
n = 18  # number of terms to print after the starting 0 and 1
i = 0  # previous Fibonacci number
j = 1  # current Fibonacci number
for iteration in range(n):
    NextNumber = i + j
    print(NextNumber)
    i = j
    j = NextNumber
````

## 3. Vectorisation vs loops

Loops are very flexible, but they are often slower and require more code, making them less readable and harder to maintain.

With `numpy`, we can apply an operation to an entire array at once through **vectorisation**. This results in code that is shorter, easier to read, and faster to execute. As a general rule, it is recommended to minimize explicit loops and make use of vectorised operations whenever possible.


### i. Math operations

Vectorization greatly simplifies array manipulation with math operations. Imagine that every temperature requires a +0.5°C shift (for example, a calibration adjustment):

````python
TempC = np.array([19.5, 20.0, 21.3, 22.1, 20.4])

# loop
TempSize = np.size(TempC)
TempCorrectedC = np.zeros(TempSize)
for i in range(0, TempSize):
    TempCorrectedC[i] = TempC[i] + 0.5
print(TempCorrectedC)

# vectorisation
TempCorrectedC = TempC + 0.5
print(TempCorrectedC)
````

You can also work with multiple arrays to define a new one. E.g. with a substraction:

````python
TempMod = np.array([19.5, 20.0, 21.3, 22.1, 20.4])
TempObs = np.array([18.2, 20.3, 20.0, 24.4, 19.8])

# loop
TempDiffSize = np.size(TempMod)
TempDiff = np.zeros(TempDiffSize)
for i in range(0, TempSize):
    TempDiff[i] = TempMod[i] - TempObs[i]
print(TempDiff)

# vectorisation
TempDiffC = TempMod - TempObs
print(TempDiffC)
````

### ii. Logical operations (filtering)

Filtering with a Boolean **mask**. A comparison such as `TempC > 21.0` does not return a single yes/no answer: NumPy gives one `True` or `False` per element. Writing `TempC[HotMask]` keeps only the temperatures where the condition holds—in this example, readings strictly above 21°C.


````python
TempC = np.array([19.5, 20.0, 21.3, 22.1, 20.4, 24.8])

HotMask = TempC > 21.0
print(HotMask)

HotTempC = TempC[HotMask]
print(HotTempC)
````

As you can see, this "rough" filtering returns an array of a different size from the original one. If you want to keep the same size, the function `np.where` can be used:

````python
HotTempsC = np.where(TempC > 21.0, TempC, np.nan)
print(HotTempsC)
````

There are also functions that can be used on arrays, similar to what we did with a `for` loop in section 1:

````python
PrecipMm = np.array([0.0, 1.2, 0.4, 2.1, 0.0, 3.5])

TotalPrecipMm = np.sum(PrecipMm)
print('Total (mm):', TotalPrecipMm)

AvgPrecipMm = np.mean(PrecipMm)
print('Average (mm):', AvgPrecipMm)
````

## Exercise G (Bonus)

You are trying to get new very temperature-humidity-dependent plants in your appartment. According to the literature, conditions become risky when the air temperature rises above 28°C **or** relative humidity falls below 40%.

You measure both variables in your five rooms, this time at four times of day (morning, afternoon, evening, and night). The eight arrays are given in the next cell.

Tasks:
- Use a combination of `vectorization`, `if` condition, and `for` loop too to give the risk in each time of day (in %) 
- Calculate the best room to take

````python
TempMorning = np.array([18.2, 19.1, 18.7, 19.4, 18.0]) # [Room 1, 2, 3, 4, 5]
TempAfternoon = np.array([28.3, 29.1, 27.4, 26.8, 30.2])
TempEvening = np.array([24.1, 25.8, 26.6, 23.9, 26.1])
TempNight = np.array([16.4, 17.2, 16.8, 17.5, 16.1])
HumMorning = np.array([78.0, 75.0, 45.0, 37.0, 45.0])
HumAfternoon = np.array([42.0, 35.0, 38.0, 48.0, 33.0])
HumEvening = np.array([55.0, 48.0, 39.0, 58.0, 45.0])
HumNight = np.array([30.0, 31.0, 43.0, 51.0, 28.0])

# Solution
CountMorning = 0
CountAfternoon = 0
CountEvening = 0
CountNight = 0

NRooms = np.size(TempMorning)
RiskPerRoom = np.zeros(NRooms)  # how many time slots are risky per room

for i in range(NRooms):
    if TempMorning[i] > 28 or HumMorning[i] < 40:  # hot OR dry
        CountMorning += 1
        RiskPerRoom[i] += 1
    if TempAfternoon[i] > 28 or HumAfternoon[i] < 40:
        CountAfternoon += 1
        RiskPerRoom[i] += 1
    if TempEvening[i] > 28 or HumEvening[i] < 40:
        CountEvening += 1
        RiskPerRoom[i] += 1
    if TempNight[i] > 28 or HumNight[i] < 40:
        CountNight += 1
        RiskPerRoom[i] += 1

print('Risk morning:', CountMorning / NRooms * 100, '%')
print('Risk afternoon:', CountAfternoon / NRooms * 100, '%')
print('Risk evening:', CountEvening / NRooms * 100, '%')
print('Risk night:', CountNight / NRooms * 100, '%')
print('Best room:', np.argmin(RiskPerRoom))  # fewest risky periods
````

Tip: Calculate the risk means counting how many sites are crossing one of the threshold for each. Same with the second task but calculate for each room.
