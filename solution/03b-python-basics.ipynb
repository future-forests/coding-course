### Why use Numpy arrays?

Numpy arrays are better for science because they:

## for Loop

Coders are the laziest person on Earth and will always try to avoid the same logic task several times in a code.

That's why we use the `for` loop to iterate the same task a number of times.

A `for` loop is accompanied by an `iterator` variable and the function `range()` which produces a sequence of numbers from the beginning (first number) to the end (second number) in steps (third number - not necessary). Here is an example of how to use both


````python
for iterator in range(0, 10, 2):
    print(iterator)
````

Try to print all odd numbers with three digits

`for` loops are very useful when manipulating arrays as they allow to navigate and perform a specific task for each element of the array.

Before using a `for` loop with an array, be sure to use the `np.size` function to find the size of the array, which is needed to indicate the end of the iteration in your `range` function.


````python
SizeTempStations = np.size(TempStations)
print(SizeTempStations)
````

The `iterator` is then used to navigate through the different elements of the array


````python
for i in range(0, SizeTempStations):
    print(TempStations[i])
````

**Note**: Python relies on indentation (space at the beginning of a line) to limit the application ("lexical scope" in coders jargon) of functions. In the case of a `for` loop, you must indent (using the TAB key on your keyboard - here it's done automatically) the lines of code to be executed in the `loop` statement. Here is an example that will not work:

````python
for iterator in range(0, 10, 2):
print(iterator)
````

To find the best location to plant a temperature-sensitive species in your garden, you measure the temperature at five different locations. The numbers you read on your thermometer are: 20.4°C, 20.8°C, 21.1°C, 20.3°C, and 21.2°C. Write code that (i) stores these numbers in an array and (ii) stores a variable that gives you the temperature average of your garden. Display the average.

Your American cousin, an avid gardener, comes to visit you and doesn't understand degrees Celsius. Convert all temperature measures to degrees Fahrenheit (you have to define a new array and use a for loop). Display the average in degrees Fahrenheit.


````python
# Solution
import statistics as stat

TempGardenC = np.array([20.4, 20.8, 21.1, 20.3, 21.2])

AvgTempC = stat.mean(TempGardenC)
print("Average (°C):", AvgTempC)

TempGardenF = np.zeros(np.size(TempGardenC))
for i in range(np.size(TempGardenC)):
    TempGardenF[i] = TempGardenC[i] * 9/5 + 32

AvgTempF = stat.mean(TempGardenF)
print("Average (°F):", AvgTempF)
````

## Condition

**New variable:** `Boolean` stores only two possible values - True or False:


````python
TempStat1 = 10.5
TempStat2 = 13.0

StationActive = True
print(StationActive)   # True
````

`Booleans` represent yes/no questions: is something true, or not? They are often the result of a `comparison` between two values. To compare values, Python uses `comparison operators`:

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |

> **Reminder:** `=` assigns a value to a variable (as you saw with `TempStat1 = 10.5`). The real "equals to" symbol is `==` (two equal signs).

When you use a comparison operator, Python returns `True` or `False`.

Here is an example:


````python
print(TempStat2 > TempStat1)  # Is station 2 warmer than station 1?
````

````python
TwoWarmerThanOne = TempStat2 > TempStat1   # Can also be assigned to a variable
print(TwoWarmerThanOne)
````

### `if` condition

These `booleans` can be used with `if` statements to **make decisions in your code**. It will execute the code in the statement only when the expression is true:

```py
if boolean:
    statement
```


````python
if TempStat2 > TempStat1:  # (TempStat2 > TempStat1) returns a True Boolean
    print(TempStat2, 'is higher than', TempStat1)
````

````python
TwoWarmerThanOne = TempStat2 > TempStat1
if TwoWarmerThanOne:  # Can also use Boolean variables directly
    print(TempStat2, 'is higher than', TempStat1)
````

You can add `else` to specify what should happen when the `if` condition is `False`.

```py
if expression:
    statement_if_true
else:
    statement_if_false
```


````python
if TempStat1 > TempStat2:
    print(TempStat2, 'is higher than', TempStat1)
else:
    print(TempStat2, 'is not higher than', TempStat1)
````

You can also combine a `for` loop with an `if` condition


````python
for i in range(0, SizeTempStations):
    if TempStations[i] > 10:
        print(TempStations[i], "°C is above 10°C")
````

## Exercise 2

An artist wants to find different places in Greenland to make ice statues of Trump. The artist needs cold conditions (below -5°C) to carve the ice. You have a database with the temperature of 20 stations where the artist can perform. In the next cell, write code that

- Gives the station number (from 0 to 19) where the artist can perform.
- Gives the temperature of the station
- Bonus: Gives the total number of stations to know how much equipment he has to bring


````python
# Solution
TrumpStations = np.array([-2.3, -4.3, -9.2, -5.0, -8.0, -2.0, -13.2, -9.0, -4.2, -9.1, -12.2, -20.8, -4.2, -3.0, -5.1, -9.2, -12.2, -5.4, -9.2, -9.3])

CountStations = 0
for i in range(np.size(TrumpStations)):
    if TrumpStations[i] < -5:
        print("Station", i, ":", TrumpStations[i], "°C")
        CountStations += 1

print("Total stations:", CountStations)
````

## Exercise 3

You wish to continue your search for the ideal location for your temperature-sensitive plant species in your garden. Therefore, you measure the temperature at the same five locations but for four different times in the day (morning, noon, afternoon and evening). Store each time of day in a separate array:

- Morning: 15.2°C, 15.4°C, 16.2°C, 16.0°C, 15.3°C
- Noon: 20.4°C, 20.8°C, 21.1°C, 20.3°C, 21.2°C
- Afternoon: 19.1°C, 18.7°C, 20.1°C, 19.9°C, 18.9°C
- Evening: 17.4°C, 16.9°C, 17.3°C, 16.8°C, 16.7°C

Write code that (i) stores these numbers in four arrays and (ii) calculates the daily average for each location and stores the results in a new array of five elements (create this array with `np.zeros`). Display the daily averages for all locations.

Bonus: You can solve this exercise with and without a for loop. Try both.


````python
# Solution
Morning = np.array([15.2, 15.4, 16.2, 16.0, 15.3])
Noon = np.array([20.4, 20.8, 21.1, 20.3, 21.2])
Afternoon = np.array([19.1, 18.7, 20.1, 19.9, 18.9])
Evening = np.array([17.4, 16.9, 17.3, 16.8, 16.7])

# Version 1: with a for loop
DailyAvg = np.zeros(np.size(Morning))
for i in range(np.size(Morning)):
    DailyAvg[i] = (Morning[i] + Noon[i] + Afternoon[i] + Evening[i]) / 4

print("Daily averages (°C):", DailyAvg)

# Version 2: without a for loop
DailyAvgNoLoop = (Morning + Noon + Afternoon + Evening) / 4
print("Daily averages (°C):", DailyAvgNoLoop)
````

## Bonus Exercise: The Fibonacci sequence

The **Fibonacci sequence** is a famous series of numbers. It starts with `0` and `1`, and each next number is the **sum of the two previous ones**:

`1, 2, 3, 5, 8, 13, 21, ...`

Write code that prints the **first 20 Fibonacci numbers**

````python
# Solution
n = 18
i = 0
j = 1

for _ in range(n):
    NextNumber = i + j
    print(NextNumber)
    i = j
    i = NextNumber
````
